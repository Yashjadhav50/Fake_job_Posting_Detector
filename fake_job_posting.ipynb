```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
fj = pd.read_csv(r"C:\Users\YASH\OneDrive\Desktop\Classes Study Material\Streamlit Projects\Fake_job_posting\fake_job_postings.csv")
fj.head(1)
```

```python
fj.shape
```

```python
fj.info()
```

```python
fj.isnull().sum()
```

```python
text_columns = ["title", "company_profile","description","requirements","benefits"]
```

```python
for col in text_columns:
    fj[col] = fj[col].fillna("")
```

```python
fj["text"] = (fj["title"] + " " + fj["company_profile"] + " " + fj["description"] + " " + fj["requirements"] + " " + fj["benefits"])
```

```python
fj[["text", "fraudulent"]].head()
```

```python
X = fj["text"]
Y = fj["fraudulent"]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=.2, random_state=1234, stratify= Y)
print("Training data :", X_train.shape)
print("Test data :", X_test.shape)
```

```python
vectorizer = TfidfVectorizer(max_features=10000, stop_words="english")
```

```python
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)
print(X_train_tfidf.shape)
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(max_iter=1000, random_state=1234, class_weight='balanced')
lr.fit(X_train_tfidf, Y_train)
```

```python
Y_pred = lr.predict(X_test_tfidf)
print("Accuracy:", accuracy_score(Y_test, Y_pred))
```

```python
print(classification_report(Y_test, Y_pred))
```

```python
import pickle

with open("fake_job_model.pkl", "wb") as file:
    pickle.dump(lr, file)

with open("tfidf_vectorizer.pkl", "wb") as file:
    pickle.dump(vectorizer, file)

print("Model saved successfully!")
print("Vectorizer saved successfully!")
```
